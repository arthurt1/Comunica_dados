```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/arthurt1/Comunica_dados/refs/heads/main/labeled-hadoop-bug-fix-dataset%20(1).csv", sep = ";")
```

```python
for coluna in df.columns:
    print(f"--- Frequência da coluna: {coluna} ---")
    print(df[coluna].value_counts())
    print("\n")
```

# Análise Exploratória
### -- Achados interessantes:

- Maioria dos bugs reportados são erros graves (Major)
- Maioria dos bugs não foi resolvido
- A maioria dos bugs relatados não foi de tratamento de exceções

```python
cont_afversions = df['AffectsVersions'].value_counts()

top_10 = cont_afversions.head(10)
top_10 = top_10.iloc[::-1]
top_10.plot(kind="barh", color="royalblue", edgecolor="black", figsize=(10, 6))

plt.title("Versões mais afetadas por bugs")
plt.xlabel("Quantidade de bugs")
plt.ylabel("Versões")
plt.grid(axis="x", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
cont_status = df['Status'].value_counts()
cont_status.plot(kind="bar", color="royalblue", edgecolor="black", figsize=(10, 6))

plt.title("Situação dos Bugs")
plt.xlabel("Status")
plt.xticks(rotation = 0)
plt.ylabel("Quantidade de bugs")

plt.tight_layout()
plt.show()
```

```python
cont_type = df['Type'].value_counts()
cont_type = cont_type.rename(index = { 0.0 : "Não Relacionado a EH", 1.0 : "Relacionado a EH"})

cont_type.plot(kind="bar", color="royalblue", edgecolor="black", figsize=(10, 6))

plt.title("Bugs de tratamento de Exceção (EH)")
plt.xlabel("")
plt.xticks(rotation = 0)
plt.ylabel("Quantidade de bugs")

plt.tight_layout()
plt.show()
```

```python
cont_priority = df["Priority"].value_counts()
cont_priority = cont_priority.iloc[::-1]

cont_priority.plot(kind="barh", color="royalblue", edgecolor="black", figsize=(12, 6))

plt.title("Bugs de tratamento de Exceção (EH)")
plt.xlabel("Quantidade de bugs")
plt.ylabel("Prioridade do Bug")

plt.tight_layout()
plt.show()
```
