```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/arthurt1/Comunica_dados/refs/heads/main/labeled-hadoop-bug-fix-dataset%20(1).csv", sep = ";")
```

```python
for coluna in df.columns:
    print(f"--- Frequência da coluna: {coluna} ---")
    print(df[coluna].value_counts())
    print("\n")
```

## Achados interessantes:

- Maioria dos bugs reportados são erros graves (Major)
- Maioria dos bugs não foi resolvido
- A maioria dos bugs relatados não foi de tratamento de exceções

```python
cont_afversions = df['AffectsVersions'].value_counts()

top_10 = cont_afversions.head(10)
top_10 = top_10.iloc[::-1]
top_10.plot(kind="barh", color="royalblue", edgecolor="black", figsize=(10, 6))

plt.title("Versões mais afetadas por bugs")
plt.xlabel("Quantidade de bugs")
plt.ylabel("Versões")
plt.grid(axis="x", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
cont_nocommits = df['NoCommits'].value_counts()

top_10 = cont_nocommits.head(10)
top_10 = top_10.iloc[::-1]
top_10.plot(kind="barh", color="royalblue", edgecolor="black", figsize=(10, 6))

plt.title("Bugs com mais commits")
plt.xlabel("Quantidades de Commits")
plt.ylabel("Bug")
plt.grid(axis="x", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```
